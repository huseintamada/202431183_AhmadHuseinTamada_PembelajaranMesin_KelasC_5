```python
db_clean = df.dropna(subset=["umur", "gaji", "tinggi", "berat", "botak_prob"])

x = db_clean[["umur", "gaji", "tinggi", "berat"]]
y = db_clean["botak_prob"]

#202431183_AhmadHuseinTamada
```

```python
#memisahkan data menjadi data latih dan data test

from sklearn.model_selection import train_test_split as tts

x_train, x_test, y_train, y_test = tts(x, y, test_size=0.2, random_state=0)

#202431183_AhmadHuseinTamada
```

```python
model = LinearRegression()

model.fit(x_train, y_train)

#202431183_AhmadHuseinTamada
```

```python
#prediksi

y_pred = model.predict(x_test)

#202431183_AhmadHuseinTamada
```

```python
#evaluasi

from sklearn.metrics import mean_squared_error, r2_score

print("Cofficient: ", model.coef_)
print("Intercept: ", model.intercept_)
print("MSE: ", mean_squared_error(y_test, y_pred))
print("Intercept: ", r2_score(y_test, y_pred))

#202431183_AhmadHuseinTamada
```

```python
#visualisasi model

plt.figure(figsize=(8, 6))
plt.scatter(y_test, y_pred, alpha=0.5)
plt.xlabel("Nilai asli y_test")
plt.ylabel("Nilai prediksi y_pred")
plt.title("Nilai asli vs Nilai prediksi")
plt.plot([0, 1], [0, 1], 'r--')
plt.show()

#202431183_AhmadHuseinTamada
```
